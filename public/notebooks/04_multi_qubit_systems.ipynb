## 💡 Concept


A single qubit's state lives in a 2-entry vector. Two independent qubits together live in a **4-entry** vector: one entry for every combination of outcomes: `|00⟩`, `|01⟩`, `|10⟩`, `|11⟩`. Three qubits: 8 entries. `n` qubits: $2^n$ entries. This exponential growth is exactly why quantum computers are interesting *and* why simulating large ones on classical hardware becomes impossible fast.

The mathematical operation that combines two independent qubit states into one joint state is called the **tensor product** (written `⊗`). Conceptually: if qubit A is in state `|a⟩` and qubit B is in state `|b⟩`, and they haven't interacted, the combined system is in state `|a⟩ ⊗ |b⟩`, usually just written `|ab⟩`.

One convention detail that trips people up: Qiskit orders qubits so that **qubit 0 is the rightmost bit** in a basis state label. So the state `|10⟩` means qubit 1 is `1` and qubit 0 is `0`. Keep an eye on this throughout the course.

## 🧪 Hands-on Code


```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

qc = QuantumCircuit(2)
qc.x(0)  # qubit 0 -> |1>
# qubit 1 left alone -> |0>
print(qc.draw())

sv = Statevector(qc)
print("Statevector:", sv.data.round(3))
print("Basis state probabilities:", sv.probabilities_dict())
```

Qubit 0 is `|1⟩`, qubit 1 is `|0⟩`. Qiskit reports the joint state as `|01⟩`, read
right-to-left: qubit 0 is the rightmost `1`, qubit 1 is the leftmost `0`. This
matches the convention described above.


## 🔁 Experiments


Let's confirm the tensor product by hand, using plain NumPy, and check it matches
what Qiskit reports.


```python
# Manually combine two independent single-qubit states with NumPy's kron (tensor product).
ket0 = np.array([1, 0])
ket1 = np.array([0, 1])

# qubit 1 = |0>, qubit 0 = |1>  ->  joint state should be |01>
manual_joint = np.kron(ket0, ket1)   # order: (qubit 1) ⊗ (qubit 0)
print("Manual tensor product:", manual_joint)

qc2 = QuantumCircuit(2)
qc2.x(0)
print("Qiskit statevector:   ", Statevector(qc2).data.real.astype(int))
```

They match. As long as qubits haven't interacted, their joint state is simply the tensor product of their individual states; no surprises yet. That "as long as they haven't interacted" clause is doing a lot of work, though. In Notebook 5, you'll build a state that **cannot** be written as any tensor product at all; that's entanglement, and it's where multi-qubit systems stop being just bookkeeping.

```python
# Now build a circuit where BOTH qubits are put into independent superpositions.
qc3 = QuantumCircuit(2)
qc3.h(0)
qc3.h(1)
print(qc3.draw())

sv3 = Statevector(qc3)
print("Probabilities:", sv3.probabilities_dict())
```

Two independent 50/50 qubits combine into **four** equally likely outcomes ( `00`,
`01`, `10`, `11` ) each at 25%. That's `0.5 × 0.5` for every combination, exactly
what you'd expect from two independent coin flips.
