## 💡 Concept


**What is Qiskit?**

Qiskit is an open-source software development kit (SDK) built for working with quantum
computers. It lets you describe a quantum circuit in Python, run it on a simulator (or
real quantum hardware), and inspect the results, all without needing to derive the
underlying physics by hand every time.

Think of it the way you'd think of a flight simulator cockpit: it doesn't change the
laws of aerodynamics, but it gives you the controls to fly.

**What is a Jupyter Notebook?**

This document you're reading *is* a Jupyter Notebook. It mixes two kinds of cells:

- **Markdown cells** (like this one) — text, explanations, images.
- **Code cells** — real, runnable Python. Click a code cell and press
  `Shift + Enter` to run it.

## 🧪 Hands-on Code


Now let's import the tools you'll reuse throughout the entire course, and build your
very first quantum circuit: a single qubit, doing nothing yet.


```python
# The core building block of every notebook in this course.
from qiskit import QuantumCircuit

# One qubit, no classical bits yet, just a container for quantum operations.
qc = QuantumCircuit(1)

# Draw it as plain text (works everywhere, no extra setup needed).
print(qc.draw())
```

An empty circuit, one horizontal wire, representing one qubit, waiting for
instructions. Let's give it one.


```python
# Add a Hadamard gate — you'll learn exactly what this does in Notebook 2.
# For now, just notice that you added an operation to the circuit.
qc.h(0)

# Draw it again, this time as a rendered diagram.
qc.draw("mpl")
```

That's it. You just built your first quantum circuit and rendered it as a diagram.
The box labeled **H** is a gate acting on your qubit. Over the next notebooks
you'll learn exactly what it does, how to measure the result, and how to build
circuits with many qubits working together.
