## 💡 Concept


In Notebook 4, every multi-qubit state you built could be described qubit-by-qubit
and then combined with a tensor product. **Entanglement** is what happens when that
stops being possible: when the joint state of two qubits carries information that
cannot be split apart and assigned to each qubit individually.

The simplest entangled state is a **Bell state**:

$$
|\Phi^+\rangle = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)
$$

Read this carefully: it says the two qubits are **either both 0 or both 1**,
with 50% probability each, and nothing in between. Measure one, and you instantly know
the other, even though neither qubit had a definite value before you measured.

## 📊 Visualization


Building a Bell pair takes exactly two gates: a Hadamard to create superposition on
one qubit, then a **CNOT** (controlled-X) to link the second qubit to it.


## 🧪 Hands-on Code


```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(2)
qc.h(0)      # put qubit 0 into superposition
qc.cx(0, 1)  # flip qubit 1 ONLY IF qubit 0 measures as 1 (a quantum "if")
print(qc.draw())

sv = Statevector(qc)
print("Statevector:", sv.data.round(3))
print("Probabilities:", sv.probabilities_dict())
```

Exactly the Bell state from the Concept section: `|00⟩` and `|11⟩`, each at 50%,
with `|01⟩` and `|10⟩` at exactly zero. The two qubits are linked. They were never a mix.


## 📐 Math Lens


Here's the proof that this state is genuinely entangled, not just a relabeled
tensor product. **Try** to write it as

$(\alpha_0|0\rangle + \beta_0|1\rangle) \otimes (\alpha_1|0\rangle + \beta_1|1\rangle)$

 and expand:

$$
\alpha_0\alpha_1|00\rangle + \alpha_0\beta_1|01\rangle + \beta_0\alpha_1|10\rangle + \beta_0\beta_1|11\rangle
$$

For this to match

 $\frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$,

 you'd need the
`|01⟩` and `|10⟩` coefficients to be exactly zero; meaning $\alpha_0\beta_1 = 0$
and $\beta_0\alpha_1 = 0$; while the `|00⟩` and `|11⟩` coefficients are both
nonzero. Try satisfying all four conditions at once with real choices of

$\alpha_0,\beta_0,\alpha_1,\beta_1$:

 you can't. There is no way to pick individual
single-qubit states for qubit 0 and qubit 1 that multiply out to the Bell state.
That impossibility **is** entanglement, expressed as algebra rather than
philosophy.

## 🔁 Experiments


Let's confirm this correlation holds under real measurement statistics, not just in
the exact statevector.


```python
from qiskit_aer import AerSimulator
from qiskit import transpile
from qiskit.visualization import plot_histogram

qc_measured = QuantumCircuit(2, 2)
qc_measured.h(0)
qc_measured.cx(0, 1)
qc_measured.measure([0, 1], [0, 1])

backend = AerSimulator()
tqc = transpile(qc_measured, backend)
counts = backend.run(tqc, shots=2000).result().get_counts()
print(counts)
plot_histogram(counts, title="Bell pair — 2000 shots")
```

Only `00` and `11` show up (within ordinary shot noise); `01` and `10` never
appear. Every single shot, the two qubits agreed with each other.

Both circuits produce a 50/50 split on each qubit individually; if you only ever
looked at one qubit at a time, you could never tell them apart. The difference only
shows up when you look at the **joint** outcomes. Entanglement isn't about either
