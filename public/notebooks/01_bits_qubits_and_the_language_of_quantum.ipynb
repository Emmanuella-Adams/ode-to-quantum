## 💡 Concept


**Classical bits.**

A classical bit is the smallest unit of classical information. It is always in
exactly one of two states: `0` or `1`. A byte is 8 bits, and every number, letter,
image, or video on a classical computer is ultimately binary. Long strings of
0s and 1s.

**Qubits.**

A qubit (quantum bit) is the quantum analog of a bit but it does not have to be
strictly `0` or strictly `1`. A qubit can exist in a **superposition** of both, and it
carries this information as *amplitudes* rather than a single fixed value. You'll
build real superposition states starting in Notebook 3; for now, just hold onto the
idea that a qubit's state is *richer* than a single 0 or 1.

A second key difference: **measurement**. Reading a classical bit never changes it. You can check a light switch as many times as you like. Reading (measuring) a qubit
forces it to collapse to a definite classical outcome, `0` or `1`, and after that
collapse the superposition is gone. This is not a limitation of our instruments. It
is a fundamental feature of quantum mechanics.


## 📊 Visualization


Qiskit represents a single qubit's state as a **statevector**: a list of complex
numbers describing the qubit. Let's look at the statevector of a qubit that hasn't
been touched by any gate yet.


## 🧪 Hands-on Code


```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# A single qubit, untouched.
qc = QuantumCircuit(1)
print(qc.draw())

sv = Statevector(qc)
print("Statevector:", sv)
```

Qiskit reports `Statevector([1.+0.j, 0.+0.j])`. That's the quantum state written as
a vector: the first entry corresponds to `|0⟩`, the second to `|1⟩`. A `1` in the
first slot and `0` in the second means: **this qubit is definitely in state |0⟩**,
exactly like a classical bit set to 0.

Let's flip it with an `X` gate (you'll study this gate properly in Notebook 2) and
look again.


```python
qc.x(0)
sv = Statevector(qc)
print(qc.draw())
print("Statevector:", sv)
```

Now the vector is `[0.+0.j, 1.+0.j]`: all the "weight" moved to the `|1⟩` slot.
Right now, both states you've built are just as classical as a light switch: fully
`0` or fully `1`. The interesting part: a qubit being *partly* both at once, is
exactly what Notebook 3 (Superposition) is about. This notebook is only about
learning to read the notation before it gets more interesting.


## 📐 Math Lens

Physicists write quantum states using **ket notation**: `|0⟩` and `|1⟩` instead of
plain `0` and `1`. This is just notation (a label) but it's the standard one you'll
see in every paper, textbook, and Qiskit doc from here on, so it's worth learning now.

$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}
\qquad\qquad
|1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

A general single-qubit state is written as a combination of both:

$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$$

where $\alpha$ and $\beta$ are complex numbers called **amplitudes**. You don't need
to manipulate this algebraically yet, just recognize the shape. You'll use it for
real starting in Notebook 3.

## 🪞 Reflection


A classical bit is a single fact: 0 or 1, full stop. A qubit's state is a *vector*:
it carries more information than a single fact, right up until the moment you
measure it. Sit with that asymmetry for a moment: **more information while
unobserved, but only one bit of it survives observation.** That tension is the
entire reason quantum computing is a different discipline, not just a faster version
of the one you already know.


## 📦 Summary

- A classical bit is always 0 or 1; a qubit's state is a vector that can hold more
- Qiskit reports a qubit's state as a `Statevector`, e.g. `[1, 0]` for |0⟩
- Measuring a qubit collapses it to a classical 0 or 1 and destroys the rest of the information
- Ket notation (`|0⟩`, `|1⟩`) is standard vocabulary you'll use for the rest of the course
